```python
from rdflib import Graph, Namespace, RDF, RDFS, OWL, URIRef

# Création du graphe
g = Graph()

# Namespace
EX = Namespace("http://example.org/university#")
g.bind("ex", EX)
g.bind("owl", OWL)
g.bind("rdfs", RDFS)

# =========================
# Classes
# =========================
classes = [
    "Course",
    "LaboratoryCourse",
    "Homework",
    "Teacher",
    "Professor",
    "Assistant"
]

for c in classes:
    g.add((EX[c], RDF.type, OWL.Class))

# Héritage
g.add((EX.LaboratoryCourse, RDFS.subClassOf, EX.Course))
g.add((EX.Professor, RDFS.subClassOf, EX.Teacher))
g.add((EX.Assistant, RDFS.subClassOf, EX.Teacher))

# Disjoint
g.add((EX.Professor, OWL.disjointWith, EX.Assistant))

# =========================
# Object Properties
# =========================

# teaches
g.add((EX.teaches, RDF.type, OWL.ObjectProperty))
g.add((EX.teaches, RDFS.domain, EX.Teacher))
g.add((EX.teaches, RDFS.range, EX.Course))

# hasHomework
g.add((EX.hasHomework, RDF.type, OWL.ObjectProperty))
g.add((EX.hasHomework, RDFS.domain, EX.Course))
g.add((EX.hasHomework, RDFS.range, EX.Homework))

# organizedBy
g.add((EX.organizedBy, RDF.type, OWL.ObjectProperty))
g.add((EX.organizedBy, RDFS.domain, EX.Course))
g.add((EX.organizedBy, RDFS.range, EX.Teacher))

# =========================
# Restrictions (OWL)
# =========================

# Professor teaches only Course
restriction_prof = URIRef(EX + "ProfessorTeachesCourse")
g.add((restriction_prof, RDF.type, OWL.Restriction))
g.add((restriction_prof, OWL.onProperty, EX.teaches))
g.add((restriction_prof, OWL.allValuesFrom, EX.Course))
g.add((EX.Professor, RDFS.subClassOf, restriction_prof))

# Assistant teaches only LaboratoryCourse
restriction_ass = URIRef(EX + "AssistantTeachesLabCourse")
g.add((restriction_ass, RDF.type, OWL.Restriction))
g.add((restriction_ass, OWL.onProperty, EX.teaches))
g.add((restriction_ass, OWL.allValuesFrom, EX.LaboratoryCourse))
g.add((EX.Assistant, RDFS.subClassOf, restriction_ass))

# =========================
# Sauvegarde en .ttl
# =========================
g.serialize(destination="universite.ttl", format="turtle")

```
